# 3.1 1D NumPy Arrays

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
```

```python
arr = np.array(list)
```

```python
multiplied = arr *4
print("Multiplied by 4:", multiplied)
```

```python
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard deviation:", np.std(arr))
```

```python
min_index = np.argmin(arr)
max_index = np.argmax(arr)
print("Index of Smallest:", min_index)
print("Index of Largest:", max_index)
```

```python
print("Value at min index:", arr[min_index])
print("Value at max index:", arr[max_index])
```

```python
print("np.min():", np.min(arr))
print("np.max():", np.max(arr))
print("Do they match?", arr[min_index] == np.min(arr) and arr[max_index] == np.max(arr))
```

```python
greater_than_20 = arr[arr > 20]
print("Values greater than 20:", greater_than_20)
```

# 2D NumPy Arrays

```python
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])
```

```python
print("Shape:", matrix.shape)
print("Rows:", matrix.shape[0])
print("Columns:", matrix.shape[1])
```

```python
print("Third column:", matrix[:, 2])
print("Second row:", matrix[1, :])
```

```python
print("Multiplied by 2: \n", matrix*2)
```

```python
print("Max of each row:", np.max(matrix, axis=1))
print("Min of each column:", np.min(matrix, axis=0))
```

```python
flat=matrix.flatten()
print("Flattened:", flat)
print("Flattened shape:", flat.shape)
```

```python
min_flat = np.argmin(matrix)
max_flat = np.argmax(matrix)
print("Flat index of smallest:", min_flat)
print("Flat index of largest:", max_flat)
```

```python
print("Smallest at (row, col):", np.unravel_index(min_flat, matrix.shape))
print("Largest at (row,col):", np.unravel_index(max_flat, matrix.shape))
```

```python
print("Values greater than 15:", matrix[matrix>15])
```

```python
print("Sorted rows:\n", np.sort(matrix, axis=1))
```

# Prime Star Systems

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i == 0:
            return False
    return True
```

```python
def rows_with_prime(x):
    result = []
    for row in x:
        for y in row:
            if is_prime(y):
                result.append(row)
    return np.array(result)
```

```python
result = rows_with_prime(arr)
print(result)
```

# Let's Play Checkers!

```python
def checkerboard():
    board = np.zeros((8,8))
    board[::2, ::2] = 1
    board[1::2, 1::2] = 1
    return board
```

```python
def reverse_checkerboard():
    board = np.zeros((8,8))
    board[::2, 1::2]=1
    board[1::2, ::2]=1
    return board
```

```python
print(checkerboard())
print(reverse_checkerboard())
```

# The Expanding Universe

```python
def expansion(words, n):
    spaces = ' ' * n
    result = []
    for word in words:
        expanded = spaces.join(word)
        result.append(expanded)
    return np.array(result)
```

# Second-Brightest Star

```python
def secondDimmest(stars):
    sorted_rows = np.sort(stars, axis=1)
    return sorted_rows[:, -2]
stars = np.array([
    [1123, 1456, 1789, 1324, 1876],
    [1567, 1987, 1678, 1405, 1589],
    [1345, 1654, 1523, 1109, 1923],
    [1298, 1890, 1367, 1784, 1432],
    [1823, 1756, 1489, 1672, 1550]])
```

```python
print(secondDimmest(stars))
```
