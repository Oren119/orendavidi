```python
import numpy as np
import matplotlib.pyplot as plt
```

# Analyzing Imported Data

```python
data = np.loadtxt("../../course_assignments/homework8/star_data.txt", skiprows=1)
```

```python
print(data.shape)
print(f"Rows: {data.shape[0]}")
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
print(f"Min magnitude: {magnitude.min()} (row {magnitude.argmin()})")
print(f"Max magnitude: {magnitude.max()} (row {magnitude.argmax()})")
```

```python
print(f"Min temperature: {temperature.min():.2f} K (row {temperature.argmin()})")
print(f"Max temperature: {temperature.max():.2f} K (row {temperature.argmax()})")
```

# Plot and HR Diagram

```python
i_bright = magnitude.argmin()
i_dim = magnitude.argmax()
i_hot = temperature.argmax()
i_cold = temperature.argmin()
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8,6))
```

```python
plt.scatter(log_temperature, log_magnitude, s=10, label= "Stars")
plt.scatter(log_temperature[i_bright], log_magnitude[i_bright],
            color="black", marker="*", s=300, label="Brightest star")
plt.scatter(log_temperature[i_dim], log_magnitude[i_dim],
            color="purple", marker="*", s=300, label="Dimmest star")
plt.scatter(log_temperature[i_hot], log_magnitude[i_hot],
            color="green", marker="*", s=300, label="Hottest star")
plt.scatter(log_temperature[i_cold], log_magnitude[i_cold],
            color="orange", marker="*", s=300, label="Coldest star")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude (log)")
plt.show()
```
