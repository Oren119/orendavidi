```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

# Importing Data

```python
df = pd.read_csv("exoplanet_data.csv", comment="#")
print(df.shape)
print(f"Rows: {df.shape[0]}, columns: {df.shape[1]}")
print(df.columns.tolist())
```

# Reducing Data

```python
'''
Planet name -> pl_name
The name of the planet most used; the planet's name, usually it's star's name and another letter 
Host name -> hostname
The name of the host star most used; name of the star that the planet orbits
Discovery method -> discoverymethod
Method for which the planet was identified; the way astronomers detect planets
Spectral Type -> st_spectype
Classification of the host star by its spectral charactersitics; letter and number label for the star's color and temperature
Equilibrium Temperature -> pl_eqt
Planet's equilibrium temperature calcualted from the heat it receives from its star; the temperature the planet would settle at from starlight
Planet mass -> pl_bmassj
The best available mass estimate in Jupiter masses; how massive the planet is compared to Jupiter
'''
```

```python
cols = ["pl_name", "hostname", "discoverymethod", "st_spectype", "pl_eqt", "pl_bmassj"]
small = df[cols]
small = small.dropna()
```

```python
print(small.shape)
print(f"Rows: {small.shape[0]}, columns: {small.shape[1]}")
print(small.head())
print(small.tail())
print(small.columns.tolist())
```

# Analyze data

```python
counts = small["discoverymethod"].value_counts()
print(counts)
```

```python
fig, ax = plt.subplots(figsize=(8,6))
for method, group in small.groupby("discoverymethod"):
    ax.scatter(group["pl_bmassj"], group["pl_eqt"], label=method)
ax.set_title("Equilibrium Temperature vs. Planet Mass")
ax.set_xlabel("Planet Mass (Jupiter masses)")
ax.set_ylabel("Equilibrium Temperature (K)")
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
for method, group in small.groupby("discoverymethod"):
    ax.scatter(np.log(group["pl_bmassj"]), np.log(group["pl_eqt"]), label=method)
ax.set_title("Log Equilibrium Temperature vs. Log Planet Mass")
ax.set_xlabel("ln(Planet Mass) [ln(Jupiter masses)]")
ax.set_ylabel("ln(Equilibrium Temperature) [ln(K)]")
ax.legend()
plt.show()
```

```python
massive = small[small["pl_bmassj"] > 5]
idx = massive["pl_eqt"].idxmin()
coldest = small.loc[idx]
print(coldest)
print(f"Name: {coldest['pl_name']}")
print(f"Index: {idx}")
print(f"Equilibrium temperature: {coldest['pl_eqt']} K")
print(f"Mass: {coldest['pl_bmassj']} Jupiter masses")
print(f"Spectral type: {coldest['st_spectype']}")
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
for method, group in small.groupby("discoverymethod"):
    ax.scatter(np.log(group["pl_bmassj"]), np.log(group["pl_eqt"]), label=method)
ax.scatter(np.log(coldest["pl_bmassj"]), np.log(coldest["pl_eqt"]),
           color="black", marker="*", s=250, label=coldest["pl_name"])
ax.set_title("Log Equilibrium Temperature vs. Log Planet Mass")
ax.set_xlabel("ln(Planet Mass) [ln(Jupiter masses)]")
ax.set_ylabel("ln(Equilibrium Temperature) [ln(K)]")
ax.legend()
plt.show()
```
